```python
# First Response Duration
```

```python
df['First Response Duration'] = (
    df['First response time'] - df['Created time']
)

df[['Created time',
   'First response time',
   'First Response Duration']].head()
```

```python
#Convert to hours:
df['First Response Hours'] = (
    df['First Response Duration'].dt.total_seconds() / 3600
)

df[['First Response Duration',
   'First Response Hours']].head()
```

```python
#Resolution duration
df['Resolution Duration'] = (
    df['Resolution time'] - df['Created time']
)

df[['Created time',
   'Resolution time',
   'Resolution Duration']].head()
```

```python
#converted to hours
df['Resolution Hours'] = (
    df['Resolution Duration'].dt.total_seconds() / 3600
)

df[['Resolution Duration',
   'Resolution Hours']].head()
```
